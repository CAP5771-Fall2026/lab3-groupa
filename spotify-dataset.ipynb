```python
# Visualization 1: Scatter Plot of Danceability vs Energy

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.scatter(
    df["danceability"],
    df["energy"],
    alpha=0.5,
    color="blue"
)

plt.title("Spotify Tracks: Danceability vs Energy")
plt.xlabel("Danceability")
plt.ylabel("Energy")

plt.grid(True, alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
# Visualization 1: Histogram of Danceability

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.hist(
    df["danceability"],
    bins=20,
    color="blue",
    edgecolor="black"
)

plt.title("Distribution of Danceability in Spotify Tracks")
plt.xlabel("Danceability")
plt.ylabel("Number of Tracks")

plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.show()
```

As shown in the histogram, most of the Spotify tracks fall within the moderate to higher danceability ranges. There are fewer tracks with very low danceability scores. Overall, this shows that a large portion of the songs in this dataset tend to have a fairly strong level of danceability.
